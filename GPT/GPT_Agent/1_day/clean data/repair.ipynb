```python
import json
import re
import os

# Kaçış karakterlerini temizleyen fonksiyon (Türkçe karakterlere zarar vermeden)
def temizle(data):
    if isinstance(data, dict):
        return {k: temizle(v) for k, v in data.items()}
    elif isinstance(data, list):
        return [temizle(v) for v in data]
    elif isinstance(data, str):
        # \u0002 karakterini temizle
        data = data.replace('\u0002', '')  # Metinde \u0002'yi temizle
        data = data.replace(chr(2), '')  # ASCII olarak \u0002'yi temizle
        data = re.sub(r'\\u0002', '', data)  # Eğer kaçış karakteri \u0002 varsa onu kaldır
        data = re.sub(r'\\[ntr]', ' ', data)  # \n, \t, \r gibi karakterleri boşluk ile değiştir
        data = re.sub(r'[\r\n\t]', ' ', data)  # \r, \n, \t karakterlerini kaldır (boşlukla değiştir)
        
        # Gereksiz boşlukları kaldır
        data = re.sub(r'\s+', ' ', data).strip()
        
        return data
    else:
        return data

# JSON dosyalarını işleyen ana fonksiyon
def process_json_files(input_folder, output_folder):
    # Klasördeki tüm dosyaları oku
    for filename in os.listdir(input_folder):
        if filename.endswith('.json'):
            input_file_path = os.path.join(input_folder, filename)
            output_file_path = os.path.join(output_folder, f"cleaned-{filename}")
            
            # JSON dosyasını oku
            with open(input_file_path, 'r', encoding='utf-8') as file:
                data = json.load(file)
            
            # Verileri temizle
            temizlenmis_data = temizle(data)
            
            # Temizlenmiş JSON dosyasını yaz
            with open(output_file_path, 'w', encoding='utf-8') as output_file:
                json.dump(temizlenmis_data, output_file, ensure_ascii=False, indent=4)
            
            print(f"Temizlenmiş dosya kaydedildi: {output_file_path}")

# Giriş ve çıkış klasör yollarını tanımla
input_folder = 'C:/Users/info/OneDrive/Masaüstü/GIT Commit Project/Gemini-ChatBot/data'
output_folder = 'C:/Users/info/OneDrive/Masaüstü/GIT Commit Project/Gemini-ChatBot/clean data'

# Çıkış klasörünü oluştur (varsa hata verme)
os.makedirs(output_folder, exist_ok=True)

# JSON dosyalarını işle
process_json_files(input_folder, output_folder)
```

```python
import json
import pandas as pd
def filter_and_sort_tours(file_path, search_term):
    with open(file_path, 'r', encoding='utf-8') as file:
        data = json.load(file)

    # Filtrelenmiş turlar listesi
    filtered_tours = []

    for item in data:
        turrip_match = search_term.lower() in item.get('turtipi', '').lower()
        kategori_ismi_match = any(search_term.lower() in kategori.get('isim', '').lower() for kategori in item.get('turKategori', []))
        
        if turrip_match or kategori_ismi_match:
            filtered_tours.append(item)

    # Puan değerlerine göre sıralama
    sorted_tours = sorted(filtered_tours, key=lambda x: max(kategori.get('puan', 0) for kategori in x.get('turKategori', [])), reverse=True)

    return sorted_tours

# Kullanım örneği
file_path = "C:/Users/info/tmps/dosya.json"
search_term = "Vizesiz Turlar"

sorted_filtered_tours = filter_and_sort_tours(file_path, search_term)
eld = pd.DataFrame(sorted_filtered_tours)
eld
```

```python
import os
import json

def clean_and_prepare_data(json_data):
    cleaned_data = []
    
    for item in json_data:
        # Kategori isimlerini ve puanlarını metin haline getir
        kategoriler = ' '.join([f"{kategori['isim']} puan: {kategori['puan']}" for kategori in item.get("turKategori", [])])
        
        # Metni ve kategorileri birleştir
        combined_text = f"{item['metin']} {kategoriler}"
        
        cleaned_item = {
            "id": item["id"],
            "isim": item["isim"],
            "turkodu": item["turkodu"],
            "gecesayisi": item["gecesayisi"],
            "geceleme": item["geceleme"],
            "konaklama": item["konaklama"],
            "ulasim": item["ulasim"],
            "ziyaretedilecekyerler": item["ziyaretedilecekyerler"],
            "vizesiz": item["vizesiz"],
            "ulasimtipi": item["ulasimtipi"],
            "turtipi": item["turtipi"],
            "combined_text": combined_text,
        }
        cleaned_data.append(cleaned_item)
    
    return cleaned_data

def process_json_files_in_directory(directory_path):
    for filename in os.listdir(directory_path):
        if filename.endswith(".json"):
            file_path = os.path.join(directory_path, filename)
            
            with open(file_path, encoding='utf-8') as f:
                tur_veri = json.load(f)
            
            cleaned_data = clean_and_prepare_data(tur_veri)

            # Temizlenmiş veriyi yeni bir dosyaya kaydet
            cleaned_filename = f"prepared_{filename}"
            cleaned_file_path = os.path.join(directory_path, cleaned_filename)
            
            with open(cleaned_file_path, 'w', encoding='utf-8') as f:
                json.dump(cleaned_data, f, indent=4, ensure_ascii=False)

            print(f"{filename} dosyası işlendi ve {cleaned_filename} olarak kaydedildi.")

# Klasör yolunu belirtin
directory_path = "d"  # Klasör yolunu buraya yazın
process_json_files_in_directory(directory_path)
```

```python
import json
import os

def merge_json_files(input_folder, output_file):
    merged_data = []

    # Klasördeki tüm JSON dosyalarını listele
    for filename in os.listdir(input_folder):
        if filename.endswith(".json"):
            file_path = os.path.join(input_folder, filename)
            
            # JSON dosyasını oku ve birleştir
            with open(file_path, "r", encoding="utf-8") as file:
                data = json.load(file)
                if isinstance(data, list):
                    merged_data.extend(data)  # Liste ise elemanları ekle
                elif isinstance(data, dict):
                    merged_data.append(data)  # Sözlük ise direkt ekle

    # Birleştirilen veriyi yeni bir JSON dosyasına kaydet
    with open(output_file, "w", encoding="utf-8") as output:
        json.dump(merged_data, output, indent=4, ensure_ascii=False)
    print(f"Birleştirilmiş JSON dosyası kaydedildi: {output_file}")

# Kullanım
input_folder = "C:/Users/info/OneDrive/Masaüstü/GIT Commit Project/Gemini-ChatBot/clean data/data"  
# JSON dosyalarının bulunduğu klasör
output_file = "C:/Users/info/OneDrive/Masaüstü/GIT Commit Project/Gemini-ChatBot/clean data/data/birlesik.json"  # Çıkış dosyası

merge_json_files(input_folder, output_file)
```
